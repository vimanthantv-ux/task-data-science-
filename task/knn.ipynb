```python
import numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
```

```python
data=pd.read_csv("Youtube_dataset_all_dataScience_channels.csv")
data
```

```python
m= np. array(data.loc[0:500,["Comment_Count","Views"]])
m
```

```python
y = np.array(data.loc[0:500,["Like_count"]]).reshape(-1, 1)
y
```

```python
model = KNeighborsClassifier(n_neighbors=3)
```

```python
model.fit(m,y)
```

```python
import numpy as np

comment =float(input("enter Comment_Count"))
views =float(input("enter Views"))

# Create a NumPy array with the numerical input values
new_fruit = np.array([[comment, views]])
prediction = model.predict(new_fruit)

print("Like_count", prediction[0])
```
